**Question 2:** In how many different states do the customers live in?

```python
from os.path import exists
import csv

# open file and init vars
filename = './sample_data/customers.csv'
row_count = 0
set_states = set()
# evaluate the contents of the file
with open(filename, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=';')
  for row in csvreader:
    if row_count > 0:
      set_states.add(row[4].strip().upper())
    row_count += 1
  print('The count of different states where customers live is:',len(set_states))
```

**Question 3** What is the state with most customers?

```python
from os.path import exists
import csv

# open file and init vars
filename = './sample_data/customers.csv'
dict_count_customers = {}
# evaluate the contents of the file
with open(filename, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=';')
  next(csvreader)
  for row in csvreader:
    if row[4].strip().upper() in dict_count_customers.keys():
      dict_count_customers[row[4].strip().upper()] += 1
    else:
      dict_count_customers[row[4].strip().upper()] = 1
  most_customers = max(dict_count_customers.values())

for key, value in dict_count_customers.items():
  if value == most_customers:
    print(f"The state with the most customers is '{key}' having {most_customers} customers ")
```
